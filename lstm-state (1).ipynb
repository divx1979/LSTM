```python
# Naive LSTM one char to one char map
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
from tensorflow.keras.utils import to_categorical
import numpy as np

tf.random.set_seed(7)

# data
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

# create mapping char to int and reverse(0-25)

char_to_int = dict((c, i) for i, c in enumerate(alphabet))
int_to_char = dict((i, c) for i, c in enumerate(alphabet))

# prepare the data for input to output pairs as int

seq_length = 1
dataX = []
dataY = []
for i in range(0, len(alphabet) - seq_length, 1):
    seq_in = alphabet[i:i + seq_length]
    seq_out = alphabet[i + seq_length]
    dataX.append([char_to_int[char] for char in seq_in])
    dataY.append(char_to_int[seq_out])
    print(seq_in, "->", seq_out)

# Reshape 
X = np.reshape(dataX, (len(dataX), seq_length, 1))
# normalize
X = X / float(len(alphabet))
# OHE outp
y = to_categorical(dataY)
# create model
model = Sequential()
model.add(LSTM(32, input_shape = (X.shape[1], X.shape[2])))
model.add(Dense(y.shape[1], activation = 'softmax'))
model.compile(loss = 'categorical_crossentropy', optimizer = 'adam',
            metrics = ['accuracy'])
model.fit(X, y, epochs = 500, batch_size = 1, verbose = 2)
sc1 = model.evaluate(X, y, verbose = 1)
print('Model Acc : %.2f%%' % (sc1[1] * 100))
# some model preds
for pattern in dataX:
    X = np.reshape(pattern, (1, len(pattern), 1))
    X = X / float(len(alphabet))
    pred1 = model.predict(X, verbose = 0)
    index = np.argmax(pred1)
    res1 = int_to_char[index]
    seq_in = [int_to_char[value] for value in pattern]
    print(seq_in, "->", res1)
```

**Stateful LSTM**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Input
from tensorflow.keras.utils import to_categorical

tf.random.set_seed(7)

# define dataset
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
char_to_int = {c: i for i, c in enumerate(alphabet)}
int_to_char = {i: c for i, c in enumerate(alphabet)}

# prepare input/output pairs
seq_length = 1
dataX, dataY = [], []
for i in range(len(alphabet) - seq_length):
    seq_in, seq_out = alphabet[i:i+seq_length], alphabet[i+seq_length]
    dataX.append([char_to_int[ch] for ch in seq_in])
    dataY.append(char_to_int[seq_out])
    print(seq_in, "->", seq_out)

# reshape and normalize
X = np.reshape(dataX, (len(dataX), seq_length, 1)) / float(len(alphabet))
y = to_categorical(dataY)

# build model
batch_size = 1
model = Sequential([
    Input(batch_shape=(batch_size, seq_length, 1)),
    LSTM(50, stateful=True, name="memory_layer"),
    Dense(y.shape[1], activation='softmax')
])
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# train model
for epoch in range(300):
    model.fit(X, y, epochs=1, batch_size=batch_size, verbose=0, shuffle=False)
    model.get_layer("memory_layer").reset_states()  # ✅ fixed targeting by name

# evaluate
scores = model.evaluate(X, y, batch_size=batch_size, verbose=0)
model.get_layer("memory_layer").reset_states()
print("Model Accuracy: %.2f%%" % (scores[1]*100))

# test predictions
seed = [char_to_int[alphabet[0]]]
for i in range(len(alphabet) - 1):
    x = np.reshape(seed, (1, len(seed), 1)) / float(len(alphabet))
    pred = model.predict(x, verbose=0)
    index = np.argmax(pred)
    print(int_to_char[seed[0]], "->", int_to_char[index])
    seed = [index]

# random start
letter = "K"
seed = [char_to_int[letter]]
print("New start:", letter)
for i in range(5):
    x = np.reshape(seed, (1, len(seed), 1)) / float(len(alphabet))
    pred = model.predict(x, verbose=0)
    index = np.argmax(pred)
    print(int_to_char[seed[0]], "->", int_to_char[index])
    seed = [index]

model.get_layer("memory_layer").reset_states()
```

**LSTM With Variable Length Input To One Char**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
from tensorflow.keras.utils import to_categorical

tf.random.set_seed(7)

# Define dataset
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
char_to_int = {c: i for i, c in enumerate(alphabet)}
int_to_char = {i: c for i, c in enumerate(alphabet)}

# --- CONFIG ---
seq_length = 5   # Max input length (can be shorter)
num_patterns = 1000  # Number of random sequences to generate
# ----------------

# Generate random subsequences of the alphabet
dataX, dataY = [], []
for _ in range(num_patterns):
    start = np.random.randint(0, len(alphabet) - seq_length - 1)
    end = np.random.randint(1, seq_length + 1)
    seq_in = alphabet[start:start + end]
    seq_out = alphabet[(start + end) % len(alphabet)]  # wrap around alphabet
    dataX.append([char_to_int[ch] for ch in seq_in])
    dataY.append(char_to_int[seq_out])

print(f"Sample 10 sequences:")
for i in range(10):
    print(''.join([int_to_char[x] for x in dataX[i]]), "->", int_to_char[dataY[i]])

# Pad/truncate sequences to fixed length for input shape compatibility
from tensorflow.keras.preprocessing.sequence import pad_sequences
X = pad_sequences(dataX, maxlen=seq_length, dtype='float32')
X = np.reshape(X, (len(X), seq_length, 1))
X = X / float(len(alphabet))
y = to_categorical(dataY)

# --- Model ---
model = Sequential([
    LSTM(75, input_shape=(seq_length, 1)),
    Dense(y.shape[1], activation='softmax')
])
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# --- Train ---
model.fit(X, y, epochs=100, batch_size=32, verbose=2)

# --- Evaluate ---
scores = model.evaluate(X, y, verbose=0)
print("Model Accuracy: %.2f%%" % (scores[1]*100))

# --- Test predictions ---
for letter in ['A', 'K', 'W', 'Y']:
    pattern = [char_to_int[letter]]
    x = np.reshape(pattern, (1, len(pattern), 1)) / float(len(alphabet))
    prediction = model.predict(x, verbose=0)
    index = np.argmax(prediction)
    print(f"{letter} -> {int_to_char[index]}")
```

```python
# LSTM with Variable Length Input Sequences to One Character Output
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import LSTM
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.preprocessing.sequence import pad_sequences
# fix random seed for reproducibility
np.random.seed(7)
tf.random.set_seed(7)
# define the raw dataset
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
# create mapping of characters to integers (0-25) and the reverse
char_to_int = dict((c, i) for i, c in enumerate(alphabet))
int_to_char = dict((i, c) for i, c in enumerate(alphabet))
# prepare the dataset of input to output pairs encoded as integers
num_inputs = 1000
max_len = 5
dataX = []
dataY = []
for i in range(num_inputs):
    start = np.random.randint(len(alphabet)-2)
    end = np.random.randint(start, min(start+max_len,len(alphabet)-1))
    sequence_in = alphabet[start:end+1]
    sequence_out = alphabet[end + 1]
    dataX.append([char_to_int[char] for char in sequence_in])
    dataY.append(char_to_int[sequence_out])
    print(sequence_in, '->', sequence_out)
# convert list of lists to array and pad sequences if needed
X = pad_sequences(dataX, maxlen=max_len, dtype='float32')
# reshape X to be [samples, time steps, features]
X = np.reshape(X, (X.shape[0], max_len, 1))
# normalize
X = X / float(len(alphabet))
# one-hot encode the output variable
y = to_categorical(dataY)
# create and fit the model
batch_size = 1
model = Sequential()
model.add(LSTM(32, input_shape=(X.shape[1], 1)))
model.add(Dense(y.shape[1], activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, y, epochs=500, batch_size=batch_size, verbose=2)
# summarize performance of the model
scores = model.evaluate(X, y, verbose=0)
print("Model Accuracy: %.2f%%" % (scores[1]*100))
# demonstrate some model predictions
for i in range(20):
    pattern_index = np.random.randint(len(dataX))
    pattern = dataX[pattern_index]
    x = pad_sequences([pattern], maxlen=max_len, dtype='float32')
    x = np.reshape(x, (1, max_len, 1))
    x = x / float(len(alphabet))
    prediction = model.predict(x, verbose=0)
    index = np.argmax(prediction)
    result = int_to_char[index]
    seq_in = [int_to_char[value] for value in pattern]
    print(seq_in, "->", result)
```
