```python
# LSTM (regression) for International Airline Passengers (complete)
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import os
from datetime import datetime
# ---- reproducibility ----
np.random.seed(7)
tf.random.set_seed(7)

# ---- user variables ----
DATA_PATH = "/kaggle/input/airline-pax/airline-pax.csv"  # update if needed
LOOK_BACK = 12                  # use 12 months (seasonal)
EPOCHS = 200
BATCH_SIZE = 16
FORECAST_STEPS = 12             # months to forecast after available data

# ---- helper: create dataset windows ----
def create_dataset(dataset, look_back=1):
    """
    dataset: 2D numpy array (n, 1)
    returns: X (n_samples, look_back), y (n_samples,)
    """
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        a = dataset[i:(i + look_back), 0]
        dataX.append(a)
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

# ---- load data ----
df = pd.read_csv(DATA_PATH, usecols=[1], header=0)  # expects csv with second column=Passengers
# if the file has header "Passengers" it's fine
values = df.values.astype('float32')

# ---- split train/test (time series split) ----
train_size = int(len(values) * 0.67)
train_values = values[:train_size]
test_values = values[train_size:]

# ---- scale using train scaler only (avoid leakage) ----
scaler = MinMaxScaler(feature_range=(0, 1))
scaler = scaler.fit(train_values)   # fit on train only

train_scaled = scaler.transform(train_values)
test_scaled = scaler.transform(test_values)

# ---- make windows ----
look_back = LOOK_BACK
trainX, trainY = create_dataset(train_scaled, look_back)
testX, testY = create_dataset(test_scaled, look_back)

# ---- reshape to [samples, time_steps, features] ----
trainX = trainX.reshape((trainX.shape[0], look_back, 1))
testX  = testX.reshape((testX.shape[0],  look_back, 1))

# ---- build model ----
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(look_back, 1)),
    LSTM(32),
    Dense(1)
])
model.compile(loss='mean_squared_error', optimizer='adam')
model.summary()

# ---- callbacks ----
os.makedirs("model_checkpoints", exist_ok=True)
chk_path = "model_checkpoints/best_lstm.h5"
es = EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True, verbose=1)
mc = ModelCheckpoint(chk_path, monitor='val_loss', save_best_only=True, verbose=1)

# ---- train ----
history = model.fit(
    trainX, trainY,
    validation_data=(testX, testY),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[es, mc],
    verbose=2
)

# ---- predictions ----
trainPredict = model.predict(trainX)
testPredict  = model.predict(testX)

# ---- invert scaling (remember scaler was fit on 1D values) ----
# scaler expects shape (n_samples, n_features)
trainPredict_inv = scaler.inverse_transform(trainPredict)
trainY_inv = scaler.inverse_transform(trainY.reshape(-1, 1))
testPredict_inv = scaler.inverse_transform(testPredict)
testY_inv = scaler.inverse_transform(testY.reshape(-1, 1))

# ---- compute RMSE ----
trainScore = math.sqrt(mean_squared_error(trainY_inv[:,0], trainPredict_inv[:,0]))
testScore  = math.sqrt(mean_squared_error(testY_inv[:,0],  testPredict_inv[:,0]))
print(f"Train Score: {trainScore:.2f} RMSE")
print(f"Test  Score: {testScore:.2f} RMSE")

# ---- prepare full-series arrays for plotting predictions aligned with original series ----
# combine train + test (original) for plotting
full_scaled = np.vstack((train_scaled, test_scaled))
full_inversed = scaler.inverse_transform(full_scaled)  # full series but scaled with train scaler

# create arrays filled with nan for plotting
trainPredictPlot = np.full((len(full_scaled), 1), np.nan, dtype=float)
testPredictPlot  = np.full((len(full_scaled), 1), np.nan, dtype=float)

# fill train predictions (start at index = look_back)
train_start = look_back
trainPredictPlot[train_start:train_start + len(trainPredict_inv), 0] = trainPredict_inv[:,0]

# fill test predictions: test starts at index train_size; predictions start at train_size + look_back
test_start = train_size + look_back
testPredictPlot[test_start:test_start + len(testPredict_inv), 0] = testPredict_inv[:,0]

# ---- plot actual vs predicted ----
plt.figure(figsize=(12,6))
plt.plot(full_inversed, label='Actual (inversed)', linewidth=1)
plt.plot(trainPredictPlot, label='Train Predict', linewidth=1)
plt.plot(testPredictPlot, label='Test Predict', linewidth=1)
plt.title('LSTM: Airline Passengers - Actual vs Predicted')
plt.xlabel('Time (months index)')
plt.ylabel('Passengers')
plt.legend()
plt.grid(True)
plt.show()

# ---- recursive multi-step forecast for next FORECAST_STEPS months ----
# take last available look_back scaled values from the entire dataset
last_sequence = np.vstack((train_scaled, test_scaled))[-look_back:]   # shape (look_back, 1)
forecast_scaled = []
current_sequence = last_sequence.copy()  # (look_back, 1)

for step in range(FORECAST_STEPS):
    # reshape to (1, look_back, 1)
    x_input = current_sequence.reshape((1, look_back, 1))
    yhat_scaled = model.predict(x_input)
    # append prediction
    forecast_scaled.append(yhat_scaled[0,0])
    # roll sequence and append predicted value
    current_sequence = np.roll(current_sequence, -1, axis=0)
    current_sequence[-1, 0] = yhat_scaled[0,0]

# invert forecast
forecast_scaled_arr = np.array(forecast_scaled).reshape(-1,1)
forecast_inv = scaler.inverse_transform(forecast_scaled_arr)

# show forecasted numbers
forecast_months = [f"t+{i+1}" for i in range(FORECAST_STEPS)]
forecast_df = pd.DataFrame({'step': forecast_months, 'forecast_passengers': forecast_inv.flatten()})
print("\nNext", FORECAST_STEPS, "months (recursive forecast):")
print(forecast_df)

# ---- plot recent history + forecast ----
plt.figure(figsize=(12,6))
# plot last 3 years of actuals for context
n_history = 36
recent_actual = full_inversed[-n_history:].flatten()
plt.plot(range(len(recent_actual)), recent_actual, label='Recent Actuals')
# plot forecast appended after recent_actual
forecast_x = np.arange(len(recent_actual), len(recent_actual) + len(forecast_inv))
plt.plot(forecast_x, forecast_inv.flatten(), marker='o', linestyle='--', label='Forecast Next 12')
plt.title('Recent Actuals and 12-step Forecast')
plt.xlabel('Time index (relative)')
plt.ylabel('Passengers')
plt.legend()
plt.grid(True)
plt.show()
```

# How To Develop LSTM Recurrent Neural Networks for Time Series Prediction in Python With Keras

```python
# import libs

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
```

```python
# load data

# Load dataset
df = pd.read_csv('/kaggle/input/airline-pax/airline-pax.csv', usecols=[1], engine='python')
dataset = df.values.astype('float32')

# Normalize (LSTMs love scaled data)
scaler = MinMaxScaler(feature_range=(0, 1))
dataset = scaler.fit_transform(dataset)

# Split into train and test
train_size = int(len(dataset) * 0.67)
test_size = len(dataset) - train_size
train, test = dataset[0:train_size,:], dataset[train_size:len(dataset),:]
```

*Create Data Matrix(Windowed Form)*

```python
def create_dataset(dataset, look_back=1):
    dataX, dataY = [], []
    for i in range(len(dataset)-look_back-1):
        a = dataset[i:(i+look_back), 0]
        dataX.append(a)
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

look_back = 3
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)
```

```python
# Now, reshape it for TensorFlow:
## LSTMs expect input in the form [samples, time steps, features]

trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 1))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 1))
```

```python
# Create a Simple LSTM for Regression

model = Sequential()
model.add(LSTM(50, input_shape=(look_back, 1)))
model.add(Dense(1))
model.compile(loss='mean_squared_error', optimizer='adam')
model.fit(trainX, trainY, epochs=100, batch_size=1, verbose=2)
```

```python
# Make Predictions and Visualize

trainPredict = model.predict(trainX)
testPredict = model.predict(testX)

# Invert normalization for plotting
trainPredict = scaler.inverse_transform(trainPredict)
trainY_inv = scaler.inverse_transform([trainY])
testPredict = scaler.inverse_transform(testPredict)
testY_inv = scaler.inverse_transform([testY])

# Root Mean Squared Error
trainScore = math.sqrt(mean_squared_error(trainY_inv[0], trainPredict[:,0]))
testScore = math.sqrt(mean_squared_error(testY_inv[0], testPredict[:,0]))
print(f'Train RMSE: {trainScore:.2f}, Test RMSE: {testScore:.2f}')
```

```python
## Plot Actual Vs Pred

trainPredictPlot = np.empty_like(dataset)
trainPredictPlot[:, :] = np.nan
trainPredictPlot[look_back:len(trainPredict)+look_back, :] = trainPredict

testPredictPlot = np.empty_like(dataset)
testPredictPlot[:, :] = np.nan
testPredictPlot[len(trainPredict)+(look_back*2)+1:len(dataset)-1, :] = testPredict

plt.figure(figsize=(12,6))
plt.plot(scaler.inverse_transform(dataset), label='Original Data')
plt.plot(trainPredictPlot, label='Train Prediction')
plt.plot(testPredictPlot, label='Test Prediction')
plt.legend()
plt.show()
```

**Time Step Formulation**

```python
# One Time Step A Time

look_back = 1
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

trainX = np.reshape(trainX, (trainX.shape[0], 1, trainX.shape[1]))
testX = np.reshape(testX, (testX.shape[0], 1, testX.shape[1]))

model = Sequential()
model.add(LSTM(50, input_shape=(1, look_back)))
model.add(Dense(1))
model.compile(loss='mean_squared_error', optimizer='adam')
model.fit(trainX, trainY, epochs=100, batch_size=1, verbose=2)
```

**Stateful LSTM**

```python
look_back = 3

# Create dataset
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, timesteps, features]
trainX = np.reshape(trainX, (trainX.shape[0], look_back, 1))
testX = np.reshape(testX, (testX.shape[0], look_back, 1))

batch_size = 1

# Define model
model = Sequential([
    LSTM(50, stateful=True, return_sequences=False),
    Dense(1)
])

# Explicitly build for stateful mode
model.build(input_shape=(batch_size, look_back, 1))

model.compile(loss='mean_squared_error', optimizer='adam')

# Train with manual reset of LSTM layer state
for i in range(100):
    print(f"Epoch {i+1}/100")
    model.fit(trainX, trainY, epochs=1, batch_size=batch_size, verbose=2, shuffle=False)
    # Reset state manually for each stateful LSTM
    for layer in model.layers:
        if hasattr(layer, 'reset_states'):
            layer.reset_states()
```

**Build a Stacked Stateful LSTM for Long Sequence Learning**

```python
look_back = 3

# Create dataset
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, timesteps, features]
trainX = np.reshape(trainX, (trainX.shape[0], look_back, 1))
testX = np.reshape(testX, (testX.shape[0], look_back, 1))

batch_size = 1

# -------------------------------
# 🚀 Define Stacked Stateful LSTM
# -------------------------------
model = Sequential([
    LSTM(50, stateful=True, return_sequences=True),
    LSTM(50, stateful=True, return_sequences=False),
    Dense(1)
])

# Build model explicitly for fixed batch size
model.build(input_shape=(batch_size, look_back, 1))

model.compile(loss='mean_squared_error', optimizer='adam')

# -------------------------------
# 🧠 Train Model Epoch-by-Epoch
# -------------------------------
for i in range(100):
    print(f"Epoch {i+1}/100")
    model.fit(trainX, trainY, epochs=1, batch_size=batch_size, verbose=2, shuffle=False)
    # Reset states for all stateful LSTM layers
    for layer in model.layers:
        if hasattr(layer, 'reset_states'):
            layer.reset_states()

# -------------------------------
# 📈 Make Predictions
# -------------------------------
trainPredict = model.predict(trainX, batch_size=batch_size)
testPredict = model.predict(testX, batch_size=batch_size)

# Invert scaling for plotting
trainPredict = scaler.inverse_transform(trainPredict)
trainY_inv = scaler.inverse_transform(trainY.reshape(-1, 1))
testPredict = scaler.inverse_transform(testPredict)
testY_inv = scaler.inverse_transform(testY.reshape(-1, 1))

# Shift train predictions for plotting
trainPredictPlot = np.empty_like(dataset)
trainPredictPlot[:, :] = np.nan
trainPredictPlot[look_back:len(trainPredict)+look_back, :] = trainPredict

# Shift test predictions for plotting (fixed)
testPredictPlot = np.empty_like(dataset)
testPredictPlot[:, :] = np.nan
test_start = len(trainPredict) + (look_back * 2)
testPredictPlot[test_start:test_start + len(testPredict), :] = testPredict

# Plot baseline and predictions
plt.figure(figsize=(10,6))
plt.plot(scaler.inverse_transform(dataset), label='Actual Data')
plt.plot(trainPredictPlot, label='Train Prediction')
plt.plot(testPredictPlot, label='Test Prediction')
plt.legend()
plt.title('LSTM Time Series Prediction - International Airline Passengers')
plt.show()
```
